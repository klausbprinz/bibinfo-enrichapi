```python
# AI generated code 


import json
from model_response import Response

# 1. Generate the schema
# We use 'mode="serialization"' to ensure it matches the final JSON output
raw_schema = Response.model_json_schema()

# 2. Extract definitions ($defs)
# Pydantic puts shared models in $defs. OpenAPI wants them in components/schemas.
definitions = raw_schema.pop("$defs", {})

# 3. Fix the internal references
# We need to change "#/$defs/..." to "#/components/schemas/..."
raw_schema_str = json.dumps(raw_schema)
definitions_str = json.dumps(definitions)

fixed_schema_str = raw_schema_str.replace("#/$defs/", "#/components/schemas/")
fixed_definitions_str = definitions_str.replace("#/$defs/", "#/components/schemas/")

# 4. Build the valid OpenAPI structure
openapi_wrapper = {
    "openapi": "3.1.0",
    "info": {
        "title": "BB Response API - OpenAPI Wrapper",
        "version": "1.0.0"
    },
    "paths": {
        "/search": {
            "get": {
                "responses": {
                    "200": {
                        "description": "Successful retrieval",
                        "content": {
                            "application/json": {
                                "schema": json.loads(fixed_schema_str)
                            }
                        }
                    }
                }
            }
        }
    },
    "components": {
        "schemas": json.loads(fixed_definitions_str)
    }
}

with open("BB_openAPIenvelope_oenbResponseSchema_v1.json", "w", encoding="utf-8") as f:
    json.dump(openapi_wrapper, f, indent=2)

print("Done")
```
