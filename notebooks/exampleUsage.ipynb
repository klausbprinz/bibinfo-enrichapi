### Full Deep Extraction (SRU, Lobid, Wikidata, Cover, Abstract, AdditionalRecs) with maxRecs

#### via BibId/AC

```python
import requests

url = "http://127.0.0.1:8000/enrich"

payload = {
    "iType": "bib",
    "institution": {
        "iName": "oenb",
        "identifier": "AC01514953",         # HERE
        "identifierType": "bibId",          # HERE -> change from ac to bibId
        
        # core features
        "fetchMarc21MD": True,                          # defaults to True
        "fetchLobidGND": True,                          # defaults to True
        "fetchWikidata": True,                          # defaults to True
        "fetchCover": True,                             # defaults to False
        "fetchDescription": True,                       # defaults to False                                          
        
        # granular sru similar recs features
        "fetchSimilarByAuthor": True,                   # defaults to False
        "fetchSimilarBySubject": True,                  # defaults to False
        "fetchSimilarByClassification": True,           # defaults to False

        # limit control for sru similar recs
        "maxRecs": 10                                   # defaults to 5, capped at 50
    }
}

res = requests.post(url, json=payload)
res.raise_for_status()

res.json()
```

```python
import requests
from requests.exceptions import HTTPError

url = "http://127.0.0.1:8000/enrich"

payload = {
    "iType": "bib",
    "institution": {
        "iName": "oenb",
        "identifier": "thisIsStupid",
        "identifierType": "bibId",                                       
    }
}

try:
    res = requests.post(url, json=payload)
    res.raise_for_status()
    print(res.json())
    
except HTTPError as e:
    # extract the custom detail message sent by FastAPI
    errorDetail = e.response.json().get("detail", "Unknown error")
    print(f"API Error ({e.response.status_code}): {errorDetail}")
```

## Example Mini-Pipes

### Read additionalRecs and call specifically (SRU only)

```python
from requests import Session

# simulate flow -> barcode first
firstRecId = "+Z168276302"

session = Session()
url = "http://127.0.0.1:8000/enrich"

# could also build a simple wrapper for API calls -> would be easy and more elegant
def fetchRec(recId, recIdType, fetchSimilar = False) -> dict:

    payload = {
        "iType": "bib",
        "institution": {
            "iName": "oenb",
            "identifier": recId,
            "identifierType": recIdType,
            "fetchMarc21MD": True,
            "fetchLobidGND": False,
            "fetchWikidata": False,
            "fetchSimilarBySubject": fetchSimilar
        }
    }

    res = requests.post(url, json=payload)
    res.raise_for_status()

    return res.json()


# fetch first via barcode
resDict = fetchRec(firstRecId, "barcode", fetchSimilar=True)
additionalRecs = resDict["response"]["result"]["additionalRecsSRU"]["records"][0]["additionalRecs"]

detailedAdditionalRecs = {}

for additionalRec in additionalRecs:
    
    # fetch additional via ACs
    detailedAdditionalRecs[additionalRec["ac"]] = fetchRec(additionalRec["ac"], "ac")

# let's have a look
detailedAdditionalRecs
```

### Read addedEntries and call specifically (Lobid and Wikidata only)

```python
# AC07527091

from requests import Session

firstRecId = "AC15037312"

session = Session()
url = "http://127.0.0.1:8000/enrich"

# could also build a simple wrapper for API calls -> would be easy and more elegant
def fetchRec(recId = None, recIdType= None, gndId = None) -> dict:

    payload = {
        "iType": "bib",
        "institution": {
            "iName": "oenb",
            "fetchLobidGND": False
        }
    }

    if recId is not None and recIdType is not None and gndId is None:
        payload["institution"]["identifier"] = recId
        payload["institution"]["identifierType"] = recIdType
        payload["institution"]["fetchMarc21MD"] = True
        payload["institution"]["fetchWikidata"] = False

    else:
        payload["institution"]["gndId"] = gndId
        payload["institution"]["fetchMarc21MD"] = False
        payload["institution"]["fetchWikidata"] = True
        payload["institution"]["fetchLobidGND"] = True

    res = requests.post(url, json=payload)
    res.raise_for_status()

    return res.json()


# base record
resDict = fetchRec(recId=firstRecId, recIdType="ac")
addedEntries = resDict["response"]["result"]["basicMarc21MD"]["addedEntries"]

addedEntryWikidataRecs = {}

for addedEntry in addedEntries:
    
    # fetch additional wikidata and lobid data
    addedEntryWikidataRecs[addedEntry["gndIdentifier"]] = fetchRec(gndId=addedEntry["gndIdentifier"])

# let's have a look
addedEntryWikidataRecs
```
